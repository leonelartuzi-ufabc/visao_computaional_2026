O código original a imagem estava em preto e branco pois o código estava com a flag 0 no trecho de código 
<code>img = cv.imread('messi5.jpg',0)</code>, se chamarmos com 1 a imagem é exibida de maneira colorida



```python
import numpy as np
import cv2 as cv

img = cv.imread('messi5.jpg',1)
cv.imshow('image',img)
k = cv.waitKey(0)
if k == 27:         # wait for ESC key to exit
    cv.destroyAllWindows()
elif k == ord('s'): # wait for 's' key to save and exit
    cv.imwrite('messigray.png',img)
    cv.destroyAllWindows()
```

Foi alterado no código o tempo do frame, ajustando o tempo e sleep da thread do python para dentro do while para chamar o próximo frame, em ret, frame = cap.read(), para deixar mais rápido, <code>time.sleep(1/120.0)</code>  e para mais lento time.

Para gerenciar o FPS do video você pode modificar a váriavel FPS, sendo que o controle do tempo de quadro é feito pelo time sleep do python

```python
import time
import numpy as np
import cv2 as cv

cap = cv.VideoCapture('big_buck_bunny.mp4')
FPS = 10

while(cap.isOpened()):
    ret, frame = cap.read()
    
    if ret==True:
        # show the frame
        cv.imshow('frame',frame)

        #wait next frame by 40ms - 25fps
        time.sleep(1/FPS) 
        
        if cv.waitKey(1) & 0xFF == ord('q'):
            break
    else:
        break
               
cap.release()
cv.destroyAllWindows()
```


Foi modificado a chamada frame = cv.flip(frame, 90) para rotacionar a imagem em 90º, também foi adicionado um time.sleep(1/30), para capturar o próximo frame, gerando um vídeo com 30fps


```python
import numpy as np
import cv2 as cv
cap = cv.VideoCapture(0)
import time


# Get current width of frame
width = cap.get(cv.CAP_PROP_FRAME_WIDTH)   # float
# Get current height of frame
height = cap.get(cv.CAP_PROP_FRAME_HEIGHT) # float
# Define Video Frame Rate in fps
fps = 10.0

# Define the codec and create VideoWriter object
fourcc = cv.VideoWriter_fourcc(*'XVID')
out = cv.VideoWriter('video1_lento_pessoa.avi', fourcc, fps, (int(width),int(height)) )

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        print("Can't receive frame (stream end?). Exiting ...")
        break
    frame = cv.flip(frame, 90)
    # write the flipped frame
    out.write(frame)
    time.sleep(1/30.0) 
        

    cv.imshow('frame', frame)
    if cv.waitKey(1) == ord('q'):
        break

# Release everything if job is finished
cap.release()
out.release()
cv.destroyAllWindows()
```
